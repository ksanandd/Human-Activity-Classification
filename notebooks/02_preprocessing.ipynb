```python
import os
import pandas as pd
import numpy as np

DATA_PATH = "../data/PAMAP2"
PROTOCOL_PATH = os.path.join(DATA_PATH, "Protocol")

print("Protocol path:", PROTOCOL_PATH)
```

```python
column_names = [
    "timestamp",
    "activity_id",
    "heart_rate",

    # Hand IMU
    "hand_temperature",
    "hand_acc_16g_x",
    "hand_acc_16g_y",
    "hand_acc_16g_z",
    "hand_acc_6g_x",
    "hand_acc_6g_y",
    "hand_acc_6g_z",
    "hand_gyro_x",
    "hand_gyro_y",
    "hand_gyro_z",
    "hand_mag_x",
    "hand_mag_y",
    "hand_mag_z",
    "hand_orientation_1",
    "hand_orientation_2",
    "hand_orientation_3",
    "hand_orientation_4",

    # Chest IMU
    "chest_temperature",
    "chest_acc_16g_x",
    "chest_acc_16g_y",
    "chest_acc_16g_z",
    "chest_acc_6g_x",
    "chest_acc_6g_y",
    "chest_acc_6g_z",
    "chest_gyro_x",
    "chest_gyro_y",
    "chest_gyro_z",
    "chest_mag_x",
    "chest_mag_y",
    "chest_mag_z",
    "chest_orientation_1",
    "chest_orientation_2",
    "chest_orientation_3",
    "chest_orientation_4",

    # Ankle IMU
    "ankle_temperature",
    "ankle_acc_16g_x",
    "ankle_acc_16g_y",
    "ankle_acc_16g_z",
    "ankle_acc_6g_x",
    "ankle_acc_6g_y",
    "ankle_acc_6g_z",
    "ankle_gyro_x",
    "ankle_gyro_y",
    "ankle_gyro_z",
    "ankle_mag_x",
    "ankle_mag_y",
    "ankle_mag_z",
    "ankle_orientation_1",
    "ankle_orientation_2",
    "ankle_orientation_3",
    "ankle_orientation_4"
]

print("Number of column names:", len(column_names))
```

```python
protocol_files = sorted([
    f for f in os.listdir(PROTOCOL_PATH)
    if f.endswith(".dat")
])

print("Number of subjects:", len(protocol_files))
print(protocol_files)
```

```python
all_subjects = []

for file in protocol_files:
    file_path = os.path.join(PROTOCOL_PATH, file)

    subject_df = pd.read_csv(
        file_path,
        sep=r"\s+",
        header=None
    )

    subject_df.columns = column_names

    subject_df["subject_id"] = file.replace(".dat", "")

    all_subjects.append(subject_df)

    print(
        f"{file}: "
        f"{subject_df.shape[0]:,} rows × "
        f"{subject_df.shape[1]} columns"
    )
```

```python
df_all = pd.concat(
    all_subjects,
    ignore_index=True
)

print("Combined dataset shape:", df_all.shape)
```

```python
print(df_all["subject_id"].value_counts().sort_index())
```

```python
activity_counts = (
    df_all["activity_id"]
    .value_counts()
    .sort_index()
)

print(activity_counts)
```

```python
df_processed = df_all.copy()
```

```python
print(
    "Activity 0 rows:",
    (df_processed["activity_id"] == 0).sum()
)
```

```python
df_processed = df_processed[
    df_processed["activity_id"] != 0
].copy()
```

```python
print("Shape after removing activity 0:", df_processed.shape)
```

```python
print(
    "Remaining activity IDs:",
    sorted(df_processed["activity_id"].unique())
)
```

```python
missing = df_processed.isna().sum()

print("Total missing values:", missing.sum())

print("\nColumns with missing values:")
print(missing[missing > 0])
```

```python
print(
    "Heart-rate missing percentage:",
    df_processed["heart_rate"].isna().mean() * 100
)
```

```python
missing_by_subject = (
    df_processed
    .groupby("subject_id")
    .apply(lambda x: x.isna().sum().sum())
)

print(missing_by_subject)
```

```python
hr_missing = df_processed["heart_rate"].isna().sum()

other_columns = [
    col for col in df_processed.columns
    if col not in ["timestamp", "activity_id", "heart_rate", "subject_id"]
]

sensor_missing = df_processed[other_columns].isna().sum().sum()

print("Missing heart-rate values:", hr_missing)
print("Missing sensor values:", sensor_missing)
```

```python
rows_with_sensor_missing = df_processed[other_columns].isna().any(axis=1)

print(
    "Rows with at least one missing sensor value:",
    rows_with_sensor_missing.sum()
)

print(
    "Percentage of rows:",
    rows_with_sensor_missing.mean() * 100
)
```

```python
sensor_groups = {
    "Hand": [c for c in df_processed.columns if c.startswith("hand_")],
    "Chest": [c for c in df_processed.columns if c.startswith("chest_")],
    "Ankle": [c for c in df_processed.columns if c.startswith("ankle_")]
}

for group, columns in sensor_groups.items():
    missing = df_processed[columns].isna().sum().sum()
    print(f"{group} missing values: {missing}")
```

```python
df_processed["heart_rate"] = (
    df_processed
    .groupby("subject_id")["heart_rate"]
    .transform(
        lambda x: x.interpolate(method="linear")
    )
)
```

```python
remaining_hr_missing = df_processed["heart_rate"].isna().sum()

print("Remaining missing heart-rate values:", remaining_hr_missing)

print(
    "Remaining percentage:",
    df_processed["heart_rate"].isna().mean() * 100
)
```

```python
remaining_hr_rows = df_processed[
    df_processed["heart_rate"].isna()
]

print(remaining_hr_rows[
    ["subject_id", "timestamp", "activity_id", "heart_rate"]
])
```

```python
remaining_hr_rows = df_processed[
    df_processed["heart_rate"].isna()
]

print(remaining_hr_rows[
    ["subject_id", "timestamp", "activity_id", "heart_rate"]
])
```

```python
for subject in remaining_hr_rows["subject_id"].unique():
    subject_data = df_processed[
        df_processed["subject_id"] == subject
    ]

    missing_indices = subject_data.index[
        subject_data["heart_rate"].isna()
    ]

    print(f"\n{subject}")
    print("Missing indices:", missing_indices.tolist())
```

```python
feature_columns = [
    col for col in df_processed.columns
    if col not in ["timestamp", "activity_id", "subject_id"]
]

complete_rows = df_processed[
    feature_columns
].notna().all(axis=1)

print("Total rows:", len(df_processed))
print("Rows with all features available:", complete_rows.sum())
print(
    "Rows with at least one missing feature:",
    (~complete_rows).sum()
)
```

```python
full_features = [
    col for col in df_processed.columns
    if col.startswith("hand_")
    or col.startswith("chest_")
    or col.startswith("ankle_")
    or col == "heart_rate"
]

print("Full feature count:", len(full_features))

full_complete = df_processed[
    full_features
].notna().all(axis=1)

print(
    "Full-feature rows with no missing values:",
    full_complete.sum()
)
```

```python
limited_features = [
    col for col in df_processed.columns
    if col.startswith("hand_")
    or col == "heart_rate"
]

print("Limited feature count:", len(limited_features))

limited_complete = df_processed[
    limited_features
].notna().all(axis=1)

print(
    "Limited-feature rows with no missing values:",
    limited_complete.sum()
)
```

```python
# Create complete datasets for both feature sets

full_data = df_processed[
    full_features + ["activity_id", "subject_id", "timestamp"]
].dropna().copy()

limited_data = df_processed[
    limited_features + ["activity_id", "subject_id", "timestamp"]
].dropna().copy()

print("Full dataset shape:", full_data.shape)
print("Limited dataset shape:", limited_data.shape)
```

```python
print("FULL DATASET")
print("Shape:", full_data.shape)
print("Activities:", sorted(full_data["activity_id"].unique()))
print("Subjects:", sorted(full_data["subject_id"].unique()))

print("\nLIMITED DATASET")
print("Shape:", limited_data.shape)
print("Activities:", sorted(limited_data["activity_id"].unique()))
print("Subjects:", sorted(limited_data["subject_id"].unique()))
```

```python
print("Full dataset activity distribution:")
print(full_data["activity_id"].value_counts().sort_index())

print("\nLimited dataset activity distribution:")
print(limited_data["activity_id"].value_counts().sort_index())
```

```python
import os

PROCESSED_PATH = "../data/processed"

os.makedirs(PROCESSED_PATH, exist_ok=True)

full_data.to_parquet(
    os.path.join(PROCESSED_PATH, "pamap2_full.parquet"),
    index=False
)

limited_data.to_parquet(
    os.path.join(PROCESSED_PATH, "pamap2_limited.parquet"),
    index=False
)

print("Datasets saved successfully.")
```
