```python
import os

DATA_PATH = "../data/PAMAP2"

print("Dataset path:", DATA_PATH)
print("\nContents:")

for item in os.listdir(DATA_PATH):
    print(item)
```

```python
protocol_path = os.path.join(DATA_PATH, "Protocol")

print("Protocol files:\n")

for item in os.listdir(protocol_path):
    print(item)
```

```python
protocol_files = [
    f for f in os.listdir(protocol_path)
    if f.endswith(".dat")
]

print("Number of protocol files:", len(protocol_files))
print("\nFiles:")
print(protocol_files)
```

```python
import pandas as pd

first_file = os.path.join(protocol_path, protocol_files[0])

df = pd.read_csv(
    first_file,
    sep=r"\s+",
    header=None
)

print("File:", protocol_files[0])
print("Shape:", df.shape)
```
